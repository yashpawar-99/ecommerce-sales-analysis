```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv("../data/train.csv")

df.shape
df
```

```python
#Total Sales by Category
total_salesby_ca = df.groupby("Category")["Sales"].sum().reset_index()

sns.barplot(x="Category",y="Sales",data=total_salesby_ca)
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.title("Total Sales by Category")
plt.show()
```

```python
#Top 5 Sub-Categories by Total Sales
sub_cat_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)
plt_sub_cat = sub_cat_sales[0:5].reset_index()

sns.barplot(x="Sub-Category",y="Sales",data=plt_sub_cat)
plt.xlabel("Sub Catagry")
plt.ylabel("Sales")
plt.title("Sub catgory by total sales")
plt.style.use("bmh")
plt.show()
```

```python
#Total Sales by Region
sales_region = df.groupby("Region")["Sales"].sum().reset_index()
sns.barplot(x="Region",y="Sales",data=sales_region)
plt.xlabel("Region")
plt.ylabel("Sales")
plt.title("Total Sales By Region")
plt.show()
```

```python
#Average Revenue per Order by Region
#avg_revanue = total_revanue / order 

order = df.groupby("Region")["Order ID"].nunique()
total_Sales= df.groupby("Region")["Sales"].sum()

avg_rev_order = (total_Sales/order)

sns.barplot(data=avg_rev_order)
plt.xlabel("Region")
plt.ylabel("Avg Revanue")
plt.title("Avrage Revanue Per Order By region")
plt.show()
```
